```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

%matplotlib inline

plt.style.use('ggplot')
```

```python
FILE_PATH = 'player_stats.csv'
YOUR_RANDOM_STATE = 5024

df_full = None
df = pd.DataFrame()

try:
    df_full = pd.read_csv(FILE_PATH, encoding='latin1')
    print(f"'{FILE_PATH}' başarıyla 'latin1' kodlaması ile okundu.")
    print(f"Toplam satır sayısı: {len(df_full)}, Sütun sayısı: {len(df_full.columns)}")

    if len(df_full) >= 2000:
        df = df_full.sample(n=2000, random_state=YOUR_RANDOM_STATE)
        print(f"2000 satır başarıyla örneklendi. Yeni DataFrame boyutu: {df.shape}")
    elif len(df_full) > 0: 
        print(f"Uyarı: Veri setinde 2000'den az satır ({len(df_full)}) bulunuyor. Tüm veri seti kullanılacak.")
        df = df_full.copy() 
    else: 
        print("Hata: Veri dosyası okundu ancak içi boş.")

except FileNotFoundError:
    print(f"Hata: '{FILE_PATH}' adlı dosya bulunamadı. Lütfen dosya adını ve yolunu kontrol edin.")
except Exception as e:
    print(f"Veri okunurken bir hata oluştu: {e}")

if not df.empty:
    print("\nÖrneklenmiş verinin ilk 5 satırı:")
    print(df.head())
else:
    print("\nVeri yükleme başarısız oldu veya veri seti boş. Lütfen yukarıdaki mesajları kontrol edin.")
```

```python
if 'value' in df.columns:
    original_value_str = df['value'].astype(str)
    
    cleaned_value_str = original_value_str.str.replace('$', '', regex=False).str.replace(',', '', regex=False).str.strip()

    def convert_value_final(val_str):
        if pd.isna(val_str) or val_str == '':
            return np.nan
        
        val_str_upper = val_str.upper()
        
        try:
            if 'M' in val_str_upper:
                return float(val_str_upper.replace('M', '')) * 1_000_000
            elif 'K' in val_str_upper:
                return float(val_str_upper.replace('K', '')) * 1_000
            else:
                # M veya K yoksa, tüm '.' karakterlerini kaldır
                return float(val_str_upper.replace('.', ''))
        except ValueError:
            return np.nan 

    df['value'] = cleaned_value_str.apply(convert_value_final)

    nan_count_after_conversion = df['value'].isnull().sum()
    if nan_count_after_conversion == 0:
        print(f"'value' sütunu başarıyla sayısala dönüştürüldü. Hiç NaN değer kalmadı.")
    else:
        print(f"Uyarı: 'value' sütunu dönüştürüldükten sonra {nan_count_after_conversion} adet NaN değer kaldı. Lütfen kontrol edin.")

    print("\nGüncellenmiş 'value' sütunu bilgileri:")
    print(df[['value']].head())
    print(df[['value']].info())
else:
    print("'value' sütunu DataFrame'de bulunamadı.")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler       
from scipy.cluster.hierarchy import dendrogram, linkage

%matplotlib inline
plt.style.use('ggplot') 
```

```python
print("--- Kümeleme İçin Veri Hazırlığı ---")

df_for_clustering = df.copy()

# Sadece sayısal sütunlar
numerical_cols_for_clustering = df_for_clustering.select_dtypes(include=np.number).columns.tolist()
df_cluster_features = df_for_clustering[numerical_cols_for_clustering].copy()
print(f"Başlangıçta seçilen sayısal özellikler: {df_cluster_features.columns.tolist()}")

#value çıkartılacak
if 'value' in df_cluster_features.columns:
    df_cluster_features = df_cluster_features.drop('value', axis=1)
    print("'value' (sınıf etiketi) kümeleme özelliklerinden çıkarıldı.")

print("\nKümeleme için son kullanılan özellikler:")
print(df_cluster_features.columns.tolist())
print(f"Özellik sayısı: {df_cluster_features.shape[1]}")
print("İlk 5 satır:")
print(df_cluster_features.head())

# Eksik değer kontrolü
if df_cluster_features.isnull().sum().any():
    print("\nUYARI: Kümeleme verisinde eksik değerler var! Lütfen kontrol edin ve doldurun.")
else:
    print("\nKümeleme verisinde eksik değer bulunmamaktadır.")
```

```python
print("\n--- Veri Standardizasyonu ---")

X_original_scale = df_cluster_features.copy()

scaler = StandardScaler()
if not X_original_scale.empty and X_original_scale.shape[0] > 1:
    X_scaled_array = scaler.fit_transform(X_original_scale)
    X_scaled_df = pd.DataFrame(X_scaled_array, columns=X_original_scale.columns, index=X_original_scale.index)

    print("Standardize edilmiş verinin ilk 5 satırı:")
    print(X_scaled_df.head())
    print("\nStandardize edilmiş verinin betimsel istatistikleri (ortalama ~0, std ~1 olmalı):")
    print(X_scaled_df.describe().round(2)) 
else:
    print("Standardizasyon için yeterli veri yok (DataFrame boş veya tek satır).")
    X_scaled_df = pd.DataFrame() 
```

```python
print("\n--- Hiyerarşik Kümeleme ve Dendrogramlar ---")

#Dendrogram
def plot_dendrogram(linked_matrix, title_suffix, p_truncate=10):
    plt.figure(figsize=(18, 7)) 
    dendrogram(linked_matrix,
               orientation='top',
               distance_sort='descending',
               truncate_mode='lastp', 
               p=p_truncate,         
               show_leaf_counts=True,
               show_contracted=True,  
               leaf_rotation=90.,      
               leaf_font_size=8.)
    plt.title(f'Hiyerarşik Kümeleme Dendrogramı ({title_suffix})', fontsize=16)
    plt.xlabel("Örnek İndeksi veya Küme Büyüklüğü (parantez içinde)", fontsize=12)
    plt.ylabel("Mesafe (Ward Linkage)", fontsize=12)
    plt.grid(axis='y')
    plt.show()

#Standardize Edilmemiş Dendrogram
print("\n1. Orijinal Ölçekli Veri ile Dendrogram:")
if not X_original_scale.empty and X_original_scale.shape[0] > 1 and X_original_scale.shape[1] > 0:
    try:
        linked_original = linkage(X_original_scale, method='ward', metric='euclidean')
        plot_dendrogram(linked_original, "Orijinal Ölçekli Veri", p_truncate=12)
    except Exception as e:
        print(f"Orijinal veri ile dendrogram oluşturulurken hata: {e}")
        print("Veri boyutları:", X_original_scale.shape)
        print("Veri içeriği (ilk 5 satır):\n", X_original_scale.head())
else:
    print("Orijinal ölçekli veri ile dendrogram çizmek için yeterli veya uygun veri yok.")
    if X_original_scale.empty: print("X_original_scale DataFrame'i boş.")
    elif X_original_scale.shape[0] <= 1: print(f"X_original_scale satır sayısı ({X_original_scale.shape[0]}) yetersiz.")
    elif X_original_scale.shape[1] == 0: print("X_original_scale'da hiç özellik (sütun) yok.")


#Standardize Edilmiş Veri ile Dendrogram
print("\n2. Standardize Edilmiş Veri ile Dendrogram:")
if not X_scaled_df.empty and X_scaled_df.shape[0] > 1 and X_scaled_df.shape[1] > 0:
    try:
        linked_scaled = linkage(X_scaled_df, method='ward', metric='euclidean')
        plot_dendrogram(linked_scaled, "Standardize Edilmiş Veri", p_truncate=12)
    except Exception as e:
        print(f"Standardize edilmiş veri ile dendrogram oluşturulurken hata: {e}")
        print("Veri boyutları:", X_scaled_df.shape)
        print("Veri içeriği (ilk 5 satır):\n", X_scaled_df.head())
else:
    print("Standardize edilmiş veri ile dendrogram çizmek için yeterli veya uygun veri yok.")
    if X_scaled_df.empty: print("X_scaled_df DataFrame'i boş.")
    elif X_scaled_df.shape[0] <= 1: print(f"X_scaled_df satır sayısı ({X_scaled_df.shape[0]}) yetersiz.")
    elif X_scaled_df.shape[1] == 0: print("X_scaled_df'de hiç özellik (sütun) yok.")
```

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.neighbors import KNeighborsRegressor
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error, r2_score
import matplotlib.pyplot as plt 
```

```python
print("--- Regresyon İçin Veri Hazırlığı ---")



y = df['value'].copy() 

X = df.select_dtypes(include=np.number).copy()
if 'value' in X.columns: 
    X = X.drop('value', axis=1)


print("\nKullanılacak Özellikler (X):")
print(X.columns.tolist())
print("X'in ilk 5 satırı:")
print(X.head())

print("\nHedef Değişken (y) ilk 5 değeri:")
print(y.head())

if X.isnull().sum().any():
    print("\nUYARI: X (özellikler) içinde eksik değerler var! Bunlar doldurulmalı.")
else:
    print("\nX (özellikler) içinde eksik değer bulunmamaktadır.")


if 'YOUR_RANDOM_STATE' not in locals() and 'YOUR_RANDOM_STATE' not in globals():
    YOUR_RANDOM_STATE = 5024
    print(f"YOUR_RANDOM_STATE '{YOUR_RANDOM_STATE}' olarak ayarlandı.")


if not X.empty and not y.empty:
    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=YOUR_RANDOM_STATE)

    print(f"\nEğitim seti boyutu: X_train shape={X_train.shape}, y_train shape={y_train.shape}")
    print(f"Test seti boyutu: X_test shape={X_test.shape}, y_test shape={y_test.shape}")
else:
    print("\nUYARI: X veya y boş olduğu için train-test split yapılamadı.")
```

```python
print("\n--- Model Eğitimi ve Değerlendirmesi ---")

model_performance = {}

if 'YOUR_RANDOM_STATE' not in locals() and 'YOUR_RANDOM_STATE' not in globals():
    YOUR_RANDOM_STATE = 5024
    print(f"YOUR_RANDOM_STATE model eğitimi için '{YOUR_RANDOM_STATE}' olarak ayarlandı.")

# 1. Lineer Regresyon
print("\n1. Lineer Regresyon...")
lr_model = LinearRegression()
lr_model.fit(X_train, y_train)
y_pred_lr = lr_model.predict(X_test)

mse_lr = mean_squared_error(y_test, y_pred_lr)
rmse_lr = np.sqrt(mse_lr)
r2_lr = r2_score(y_test, y_pred_lr)
model_performance['Linear Regression'] = {'MSE': mse_lr, 'RMSE': rmse_lr, 'R2 Score': r2_lr}
print(f"  MSE: {mse_lr:.2f}, RMSE: {rmse_lr:.2f}, R2 Score: {r2_lr:.4f}")


# 2. KNN Regresyon
print("\n2. KNN Regresyon (k=5)...")
knn_model = KNeighborsRegressor(n_neighbors=5)
knn_model.fit(X_train, y_train)
y_pred_knn = knn_model.predict(X_test)

mse_knn = mean_squared_error(y_test, y_pred_knn)
rmse_knn = np.sqrt(mse_knn)
r2_knn = r2_score(y_test, y_pred_knn)
model_performance['KNN Regression (k=5)'] = {'MSE': mse_knn, 'RMSE': rmse_knn, 'R2 Score': r2_knn}
print(f"  MSE: {mse_knn:.2f}, RMSE: {rmse_knn:.2f}, R2 Score: {r2_knn:.4f}")


# 3. Decision Tree
print("\n3. Karar Ağacı Regresyonu...")
dt_model = DecisionTreeRegressor(random_state=YOUR_RANDOM_STATE)
dt_model.fit(X_train, y_train)
y_pred_dt = dt_model.predict(X_test)

mse_dt = mean_squared_error(y_test, y_pred_dt)
rmse_dt = np.sqrt(mse_dt)
r2_dt = r2_score(y_test, y_pred_dt)
model_performance['Decision Tree Regression'] = {'MSE': mse_dt, 'RMSE': rmse_dt, 'R2 Score': r2_dt}
print(f"  MSE: {mse_dt:.2f}, RMSE: {rmse_dt:.2f}, R2 Score: {r2_dt:.4f}")


print("\n4. Random Forest Regresyonu...")
rf_model = RandomForestRegressor(random_state=YOUR_RANDOM_STATE, n_estimators=100) 
rf_model.fit(X_train, y_train)
y_pred_rf = rf_model.predict(X_test)

mse_rf = mean_squared_error(y_test, y_pred_rf)
rmse_rf = np.sqrt(mse_rf)
r2_rf = r2_score(y_test, y_pred_rf)
model_performance['Random Forest Regression'] = {'MSE': mse_rf, 'RMSE': rmse_rf, 'R2 Score': r2_rf}
print(f"  MSE: {mse_rf:.2f}, RMSE: {rmse_rf:.2f}, R2 Score: {r2_rf:.4f}")


print("\n--- Tüm Modellerin Performans Özeti ---")
performance_df = pd.DataFrame(model_performance).T 
performance_df = performance_df.sort_values(by='R2 Score', ascending=False) 
# MSE ve RMSE için formatlama
pd.options.display.float_format = '{:.2f}'.format
print(performance_df[['MSE', 'RMSE', 'R2 Score']])
pd.reset_option('display.float_format')
```

```python
print("\n--- KNN Regresyon Parametre Ayarlaması (k için en iyi değeri bulma) ---")

def find_best_k_for_knn_regressor(X_train, y_train, X_test, y_test, min_k=1, max_k=50):
    """
    Farklı k değerleri için KNN Regresyon modellerini eğitir ve test eder,
    en iyi R2 skorunu veren k değerini döndürür.
    Ayrıca R2 skorlarının k'ya göre grafiğini çizer.
    """
    k_values = list(range(min_k, max_k + 1))
    r2_scores_for_k = []
    best_k = min_k
    best_r2 = -float('inf') 

    print(f"{min_k}'dan {max_k}'a kadar k değerleri deneniyor...")
    for k in k_values:
        knn_temp_model = KNeighborsRegressor(n_neighbors=k)
        knn_temp_model.fit(X_train, y_train)
        y_pred_temp = knn_temp_model.predict(X_test)
        current_r2 = r2_score(y_test, y_pred_temp)
        r2_scores_for_k.append(current_r2)

        if current_r2 > best_r2:
            best_r2 = current_r2
            best_k = k
        
        if k % 10 == 0:
            print(f"  k={k} için R2: {current_r2:.4f}")

    print(f"\nEn iyi k değeri: {best_k} (R2 Skoru: {best_r2:.4f})")

    plt.figure(figsize=(12, 6))
    plt.plot(k_values, r2_scores_for_k, marker='o', linestyle='-')
    plt.title('KNN Regresyon: Farklı k Değerleri İçin R2 Skoru')
    plt.xlabel('Komşu Sayısı (k)')
    plt.ylabel('R2 Skoru (Test Seti Üzerinde)')
    plt.xticks(np.arange(min_k, max_k + 1, step=max(1, (max_k - min_k) // 10)))
    plt.grid(True)
    plt.axvline(best_k, color='red', linestyle='--', label=f'En İyi k = {best_k} (R2={best_r2:.4f})')
    plt.legend()
    plt.show()

    return best_k, best_r2

# Fonksiyonu çağıralım
# X_train, y_train, X_test, y_test daha önceki adımdan gelmeli
if 'X_train' in locals() and 'y_train' in locals() and 'X_test' in locals() and 'y_test' in locals():
    best_k_found, best_r2_found = find_best_k_for_knn_regressor(X_train, y_train, X_test, y_test, min_k=1, max_k=50)
    print(f"\nKNN Regresyon için fonksiyon tarafından bulunan en iyi k: {best_k_found} (R2 Skoru: {best_r2_found:.4f})")


else:
    print("UYARI: KNN parametre ayarı için X_train, y_train, X_test, y_test tanımlı değil.")
```

```python
if ('top_5_features_mi_names' in locals() and top_5_features_mi_names and
    'X_b' in locals() and 'y_b' in locals() and 'cv_splitter' in locals() and
    'YOUR_RANDOM_STATE' in locals() and 'baseline_rf_cv_accuracy_mean' in locals()):
    
    print("\n5.b Mutual Information ile Seçilen En İyi 5 Özellik ile Random Forest Değerlendirmesi:")
    
    # En iyi 5 özelliği X_b'den seç
    X_b_top5_mi = X_b[top_5_features_mi_names].copy()
    print(f"Mutual Info ile seçilen özelliklerle oluşturulan X_b_top5_mi şekli: {X_b_top5_mi.shape}")

    scaler_mi_top5 = StandardScaler()
    X_b_top5_mi_scaled = scaler_mi_top5.fit_transform(X_b_top5_mi)
    print("Seçilen 5 özellik StandardScaler ile ölçeklendirildi.")

    rf_model_mi_top5 = RandomForestClassifier(random_state=YOUR_RANDOM_STATE, n_estimators=100)

    cv_scores_rf_mi_top5 = cross_val_score(rf_model_mi_top5, X_b_top5_mi_scaled, y_b, cv=cv_splitter, scoring='accuracy')

    mean_cv_accuracy_mi_top5 = cv_scores_rf_mi_top5.mean()
    std_cv_accuracy_mi_top5 = cv_scores_rf_mi_top5.std()

    print(f"\nMutual Info ile seçilen en iyi 5 özellik kullanılarak Random Forest Ortalama CV Doğruluğu: {mean_cv_accuracy_mi_top5:.4f} +/- {std_cv_accuracy_mi_top5:.4f}")
    

    impact_mi = mean_cv_accuracy_mi_top5 - baseline_rf_cv_accuracy_mean
    print(f"Tüm özelliklere (baseline: {baseline_rf_cv_accuracy_mean:.4f}) kıyasla doğruluk değişimi (MI top 5): {impact_mi:+.4f}")
    

    if impact_mi > 0.001: 
        print("Yorum: Mutual Information ile seçilen en iyi 5 özellik, model performansını (CV accuracy) artırdı.")
    elif impact_mi < -0.001:
        print("Yorum: Mutual Information ile seçilen en iyi 5 özellik, model performansını (CV accuracy) düşürdü.")
    else:
        print("Yorum: Mutual Information ile seçilen en iyi 5 özellik, model performansını (CV accuracy) önemli ölçüde değiştirmedi.")
else:
    print("HATA: Mutual Info ile özellik seçimi sonrası RF değerlendirmesi için gerekli değişkenler eksik. Lütfen önceki hücreleri kontrol edin.")
    mean_cv_accuracy_mi_top5 = float('nan')
    impact_mi = float('nan')
```

```python

print("\n\n--- Yöntem 2: RandomForestClassifier feature_importances_ ---")

if 'X_b' in locals() and 'y_b' in locals() and not X_b.empty and 'YOUR_RANDOM_STATE' in locals():
    rf_for_importance = RandomForestClassifier(random_state=YOUR_RANDOM_STATE, n_estimators=100)
    rf_for_importance.fit(X_b, y_b)

    importances = rf_for_importance.feature_importances_
    importance_series = pd.Series(importances, index=X_b.columns, name="RF_Importances").sort_values(ascending=False)

    top_5_features_rf_names = importance_series.head(5).index.tolist()

    print("\n5.a RandomForestClassifier feature_importances_ ile En İyi 5 Özellik ve Önem Dereceleri:")
    print(importance_series.head(5))
    print(f"\nEn iyi 5 özellik (isimleri): {top_5_features_rf_names}")
else:
    print("HATA: Özellik seçimi için X_b, y_b veya YOUR_RANDOM_STATE tanımlı değil. Lütfen önceki adımları kontrol edin.")
    top_5_features_rf_names = []
```

```python
if ('top_5_features_rf_names' in locals() and top_5_features_rf_names and
    'X_b' in locals() and 'y_b' in locals() and 'cv_splitter' in locals() and
    'YOUR_RANDOM_STATE' in locals() and 'baseline_rf_cv_accuracy_mean' in locals()):
    
    print("\n5.b RF feature_importances_ ile Seçilen En İyi 5 Özellik ile Random Forest Değerlendirmesi:")

    X_b_top5_rf = X_b[top_5_features_rf_names].copy()
    print(f"RF importances ile seçilen özelliklerle oluşturulan X_b_top5_rf şekli: {X_b_top5_rf.shape}")


    scaler_rf_top5 = StandardScaler()
    X_b_top5_rf_scaled = scaler_rf_top5.fit_transform(X_b_top5_rf)
    print("Seçilen 5 özellik StandardScaler ile ölçeklendirildi.")


    rf_model_rf_top5 = RandomForestClassifier(random_state=YOUR_RANDOM_STATE, n_estimators=100)


    cv_scores_rf_rf_top5 = cross_val_score(rf_model_rf_top5, X_b_top5_rf_scaled, y_b, cv=cv_splitter, scoring='accuracy')
    
    mean_cv_accuracy_rf_top5 = cv_scores_rf_rf_top5.mean()
    std_cv_accuracy_rf_top5 = cv_scores_rf_rf_top5.std()

    print(f"\nRF feature_importances_ ile seçilen en iyi 5 özellik kullanılarak Random Forest Ortalama CV Doğruluğu: {mean_cv_accuracy_rf_top5:.4f} +/- {std_cv_accuracy_rf_top5:.4f}")


    impact_rf = mean_cv_accuracy_rf_top5 - baseline_rf_cv_accuracy_mean
    print(f"Tüm özelliklere (baseline: {baseline_rf_cv_accuracy_mean:.4f}) kıyasla doğruluk değişimi (RF top 5): {impact_rf:+.4f}")
    

    if impact_rf > 0.001:
        print("Yorum: RF feature_importances_ ile seçilen en iyi 5 özellik, model performansını (CV accuracy) artırdı.")
    elif impact_rf < -0.001:
        print("Yorum: RF feature_importances_ ile seçilen en iyi 5 özellik, model performansını (CV accuracy) düşürdü.")
    else:
        print("Yorum: RF feature_importances_ ile seçilen en iyi 5 özellik, model performansını (CV accuracy) önemli ölçüde değiştirmedi.")
else:
    print("HATA: RF feature_importances_ ile özellik seçimi sonrası RF değerlendirmesi için gerekli değişkenler eksik. Lütfen önceki hücreleri kontrol edin.")
    mean_cv_accuracy_rf_top5 = float('nan')
    impact_rf = float('nan')
```

```python


print("\n\n--- Özellik Seçimi Yöntemleri Üzerine Yorumlar ---")

if 'baseline_rf_cv_accuracy_mean' not in locals():
    print("UYARI: baseline_rf_cv_accuracy_mean tanımlı değil. Karşılaştırma yapılamıyor.")
else:
    print(f"Referans (Tüm Özellikler - X_b_scaled_for_cv) Random Forest Ortalama CV Doğruluğu: {baseline_rf_cv_accuracy_mean:.4f}")

    # Mutual Information sonuçları
    if ('top_5_features_mi_names' in locals() and top_5_features_mi_names and
        'mean_cv_accuracy_mi_top5' in locals() and 'impact_mi' in locals()):
        print(f"\n1. Mutual Information (MI) ile Seçilen En İyi 5 Özellik:")
        print(f"   - Seçilen Özellikler: {top_5_features_mi_names}")
        print(f"   - Bu özelliklerle RF Ortalama CV Doğruluğu: {mean_cv_accuracy_mi_top5:.4f} (Baseline'a göre değişim: {impact_mi:+.4f})")
    else:
        print("\n1. Mutual Information ile özellik seçimi sonuçları bulunamadı veya hesaplanamadı.")

    # RF Feature Importances sonuçları
    if ('top_5_features_rf_names' in locals() and top_5_features_rf_names and
        'mean_cv_accuracy_rf_top5' in locals() and 'impact_rf' in locals()):
        print(f"\n2. RandomForest feature_importances_ (RFI) ile Seçilen En İyi 5 Özellik:")
        print(f"   - Seçilen Özellikler: {top_5_features_rf_names}")
        print(f"   - Bu özelliklerle RF Ortalama CV Doğruluğu: {mean_cv_accuracy_rf_top5:.4f} (Baseline'a göre değişim: {impact_rf:+.4f})")
    else:
        print("\n2. RandomForest feature_importances_ ile özellik seçimi sonuçları bulunamadı veya hesaplanamadı.")

    print("\n3. Genel Değerlendirme ve Karşılaştırma:")
    # Seçilen özellik setlerinin karşılaştırılması
    if ('top_5_features_mi_names' in locals() and top_5_features_mi_names and
        'top_5_features_rf_names' in locals() and top_5_features_rf_names):
        common_features = set(top_5_features_mi_names).intersection(set(top_5_features_rf_names))
        print(f"   - İki yöntemle seçilen ortak özellik sayısı: {len(common_features)} ({list(common_features)})")
        if len(common_features) == 5 and set(top_5_features_mi_names) == set(top_5_features_rf_names):
            print("     > Her iki yöntem de birebir aynı en iyi 5 özelliği seçti.")
        elif len(common_features) > 0 :
            print("     > İki yöntem kısmen farklı özellik setleri seçti.")
        else:
            print("     > İki yöntem tamamen farklı özellik setleri seçti.")
        print("     > Yorum: Mutual Information (MI), her bir özelliğin hedefle olan ilişkisini tek başına (univariate) değerlendirir. RandomForest feature_importances_ (RFI) ise özelliklerin modelin (ağaçların) yapımına yaptığı katkıyı, dolayısıyla özellikler arası etkileşimleri de dolaylı olarak dikkate alarak önem derecesi atar. Bu temel farklılık nedeniyle farklı özellik setleri seçmeleri beklenebilir.")
    
    # Performansların karşılaştırılması
    print("\n   - Performans Etkisi:")
    print("     > Azaltılmış veri seti (5 özellik) kullanmak, model karmaşıklığını azaltır ve potansiyel olarak eğitim süresini kısaltır.")
    if ('impact_mi' in locals() and not pd.isna(impact_mi)) and ('impact_rf' in locals() and not pd.isna(impact_rf)):
        if impact_mi >= -0.005 and impact_rf >= -0.005: # Performans çok düşmediyse veya arttıysa
             print("     > Her iki yöntemle de model doğruluğunun önemli ölçüde düşmemiş olması (veya artması), çıkarılan özelliklerin gürültülü, gereksiz veya daha az bilgilendirici olduğunu ve ilk 5 özelliğin önemli bilgiyi taşıdığını düşündürmektedir.")
        else:
             print("     > En az bir yöntemde model doğruluğunun belirgin şekilde düşmesi, önemli bilgiler içeren bazı özelliklerin çıkarılmış olabileceğini düşündürür.")
    
        if mean_cv_accuracy_rf_top5 > mean_cv_accuracy_mi_top5 + 0.001: # Küçük bir fark eşiği
            print("     > RFI ile seçilen özellikler, Random Forest modeli için MI ile seçilenlere göre daha iyi performans vermiştir. Bu, özelliklerin değerlendirilecek modelin (RF) kendi iç yapısına göre seçilmesinin bir avantajı olabilir.")
        elif mean_cv_accuracy_mi_top5 > mean_cv_accuracy_rf_top5 + 0.001:
            print("     > MI ile seçilen özellikler, bu durumda RFI ile seçilenlere göre daha iyi performans vermiştir. Bu, MI'nin modelden bağımsız olarak yakaladığı bazı ilişkilerin RF için daha faydalı olduğunu gösterebilir.")
        else:
            print("     > Her iki özellik seçimi yöntemi de Random Forest modeli için benzer performanslar sunmuştur.")
    else:
        print("     > Performans karşılaştırması için gerekli veriler eksik.")
```

```python

from sklearn.decomposition import PCA

print("PCA için sklearn.decomposition.PCA kütüphanesi yüklendi.")
```

```python
#PCA

print("\n--- 6. PCA (Temel Bileşen Analizi) ---")

if ('X_b_scaled_for_cv' in locals() and X_b_scaled_for_cv is not None and 
    X_b_scaled_for_cv.shape[0] > 0 and 'YOUR_RANDOM_STATE' in locals()):
    
    n_original_features = X_b_scaled_for_cv.shape[1]
    n_components_pca = min(5, n_original_features) 

    if n_original_features == 0:
        print("HATA: X_b_scaled_for_cv içinde hiç özellik bulunmuyor. PCA uygulanamaz.")
        X_b_pca_first5 = None
        total_explained_variance_first_n = float('nan')
    else:
        print(f"Orijinal özellik sayısı: {n_original_features}")
        print(f"PCA için analiz edilecek temel bileşen sayısı: {n_components_pca}")

        pca_full = PCA(random_state=YOUR_RANDOM_STATE)
        pca_full.fit(X_b_scaled_for_cv)
        
         
        pca_first_n_components = PCA(n_components=n_components_pca, random_state=YOUR_RANDOM_STATE)
        X_b_pca_first5 = pca_first_n_components.fit_transform(X_b_scaled_for_cv)

        explained_variance_ratio_first_n = pca_first_n_components.explained_variance_ratio_ 
        cumulative_explained_variance_all = np.cumsum(pca_full.explained_variance_ratio_)
        total_explained_variance_first_n = cumulative_explained_variance_all[n_components_pca - 1]

        print(f"\nİlk {n_components_pca} Temel Bileşenin Açıkladığı Varyans Oranları:")
        for i, ratio in enumerate(explained_variance_ratio_first_n):
            print(f"  Temel Bileşen {i+1}: {ratio:.4f} (Bu bileşenin kümülatif katkısı: {cumulative_explained_variance_all[i]:.4f})")
        
        print(f"\nİlk {n_components_pca} temel bileşenin toplam açıkladığı varyans oranı: {total_explained_variance_first_n:.4f}")

        #Yorum
        print("\nPCA Sonuçları Yorumu (Açıklanan Varyans):")
        print(f"  - İlk {n_components_pca} temel bileşen, veri setindeki toplam varyansın yaklaşık %{total_explained_variance_first_n*100:.2f}'ini açıklamaktadır.")
        if total_explained_variance_first_n > 0.8:
            print("    Bu, verideki bilginin büyük bir kısmının daha düşük bir boyutta ({} bileşenle) temsil edilebildiğini gösterir.".format(n_components_pca))
        elif total_explained_variance_first_n < 0.5:
            print("    Bu, ilk {} bileşenin verideki varyansın yarısından azını açıkladığını, yani önemli miktarda bilginin bu ilk bileşenler dışında kaldığını gösterir.".format(n_components_pca))
        else:
            print("    Bu değer, ilk {} bileşenin verideki varyansın makul bir kısmını açıkladığını gösterir.".format(n_components_pca))

        plt.figure(figsize=(10, 6))
        
        num_components_to_plot = min(n_original_features, 15)
        
        plt.bar(range(1, num_components_to_plot + 1), 
                pca_full.explained_variance_ratio_[:num_components_to_plot], 
                alpha=0.7, align='center',
                label='Bireysel Açıklanan Varyans')
        plt.step(range(1, num_components_to_plot + 1), 
                 cumulative_explained_variance_all[:num_components_to_plot], 
                 where='mid',
                 label='Kümülatif Açıklanan Varyans', color='red')
                 
        plt.ylabel('Açıklanan Varyans Oranı')
        plt.xlabel('Temel Bileşen Sayısı')
        plt.title(f'İlk {num_components_to_plot} Temel Bileşen için Açıklanan Varyans')
        plt.xticks(range(1, num_components_to_plot + 1))
        plt.legend(loc='best')
        plt.grid(True)
        if n_components_pca <= num_components_to_plot :
            plt.axhline(y=total_explained_variance_first_n, color='purple', linestyle='--', 
                        label=f'İlk {n_components_pca} Bileşen Kümülatif Varyansı ({total_explained_variance_first_n:.2f})')
            plt.legend(loc='best')
        plt.show()
else:
    print("HATA: PCA uygulaması için X_b_scaled_for_cv veya YOUR_RANDOM_STATE tanımlı değil/boş. Lütfen önceki adımları kontrol edin.")
    X_b_pca_first5 = None 
    total_explained_variance_first_n = float('nan')
```

```python
# İlk 5 Temel Bileşen ile Random Forest Eğitimi ve Değerlendirmesi


if (X_b_pca_first5 is not None and 'y_b' in locals() and 'cv_splitter' in locals() and
    'YOUR_RANDOM_STATE' in locals() and 'baseline_rf_cv_accuracy_mean' in locals()):
    
    print("\n--- İlk 5 Temel Bileşen ile Random Forest Değerlendirmesi ---")
    print(f"PCA ile dönüştürülmüş (ilk 5 bileşen) özellik matrisi X_b_pca_first5 şekli: {X_b_pca_first5.shape}")


    rf_model_pca5 = RandomForestClassifier(random_state=YOUR_RANDOM_STATE, n_estimators=100)


    cv_scores_rf_pca5 = cross_val_score(rf_model_pca5, X_b_pca_first5, y_b, cv=cv_splitter, scoring='accuracy')

    mean_cv_accuracy_pca5 = cv_scores_rf_pca5.mean()
    std_cv_accuracy_pca5 = cv_scores_rf_pca5.std()

    print(f"\nİlk 5 temel bileşen kullanılarak Random Forest Ortalama CV Doğruluğu: {mean_cv_accuracy_pca5:.4f} +/- {std_cv_accuracy_pca5:.4f}")

    impact_pca = mean_cv_accuracy_pca5 - baseline_rf_cv_accuracy_mean
    print(f"Tüm orijinal özelliklere (baseline: {baseline_rf_cv_accuracy_mean:.4f}) kıyasla doğruluk değişimi (PCA ilk 5): {impact_pca:+.4f}")

    print("\nPCA Kullanımının Model Performansı Üzerine Yorumlar (Madde 6 Sonu):")
    print(f"  - Referans (tüm ölçeklenmiş orijinal özellikler) RF modeli CV doğruluğu: {baseline_rf_cv_accuracy_mean:.4f}")
    print(f"  - PCA (ilk 5 bileşen) ile RF modeli CV doğruluğu: {mean_cv_accuracy_pca5:.4f}")
    
    if impact_pca > 0.001:
        print("  > Yorum: PCA ile elde edilen ilk 5 temel bileşen, Random Forest modelinin performansını (CV accuracy) artırdı. Bu, PCA'nın gürültüyü azaltarak veya daha anlamlı bir özellik uzayı bularak fayda sağladığını ve bu yeni uzayın RF için daha iyi olduğunu gösterebilir.")
    elif impact_pca < -0.01: 
        print("  > Yorum: PCA ile elde edilen ilk 5 temel bileşen, Random Forest modelinin performansını (CV accuracy) düşürdü. Bu durum, ilk 5 bileşenin verideki varyansın önemli bir kısmını açıklamasına rağmen (eğer açıkladıysa), sınıflandırma için kritik olan bazı bilgilerin bu boyut azaltma işlemi sırasında kaybedildiğini gösterebilir.")
    else:
        print("  > Yorum: PCA ile elde edilen ilk 5 temel bileşen, Random Forest modelinin performansını (CV accuracy) önemli ölçüde değiştirmedi. Bu, boyut azaltmanın performanstan büyük bir ödün vermeden yapılabileceğini veya 5 bileşenin sınıflandırma için yeterli ayırt edici bilgiyi taşıdığını gösterebilir.")
    
    if 'total_explained_variance_first_n' in locals() and not pd.isna(total_explained_variance_first_n):
        print(f"  - Hatırlatma: İlk 5 temel bileşen, toplam varyansın yaklaşık %{total_explained_variance_first_n*100:.2f}'ini açıklıyordu.")
        print("    Bu açıklanan varyans oranı ile model performansı arasındaki ilişki önemlidir. Eğer varyansın büyük bir kısmı (%80-90+) açıklanmasına rağmen performans düşüyorsa, kalan küçük varyansın sınıflandırma için önemli olduğu veya PCA'nın oluşturduğu bileşenlerin RF için ideal olmadığı düşünülebilir.")

    print("  - Genel olarak PCA, özellikler arasındaki korelasyonu ortadan kaldırır ve boyutu azaltır. Bu, bazı modeller için faydalı olabilir ve 'curse of dimensionality' (boyutluluk laneti) etkisini azaltabilir.")
    print("  - Ancak, temel bileşenler orijinal özelliklerin lineer kombinasyonları olduğu için yorumlanabilirlikleri azalır. Yani, bir temel bileşenin iş dünyası açısından ne anlama geldiğini söylemek zorlaşır.")
    
    print("\n  - Diğer Yöntemlerle Karşılaştırma:")
    if 'mean_cv_accuracy_rf_top5' in locals() and not pd.isna(mean_cv_accuracy_rf_top5):
        print(f"    > RF feature_importances_ (ilk 5 özellik) ile RF CV doğruluğu: {mean_cv_accuracy_rf_top5:.4f}")
    if 'mean_cv_accuracy_mi_top5' in locals() and not pd.isna(mean_cv_accuracy_mi_top5):
        print(f"    > Mutual Info (ilk 5 özellik) ile RF CV doğruluğu: {mean_cv_accuracy_mi_top5:.4f}")


else:
    print("HATA: PCA ile dönüştürülmüş X_b_pca_first5 bulunamadığı veya diğer gerekli değişkenler eksik olduğu için bu adım atlanıyor.")
```
